# 3. COVID-19 Data Analysis
Project Question:
 Analyze COVID-19 data to understand the trends in cases, recoveries, and deaths over time.
Tasks:
Load COVID-19 case data using Pandas.
Handle missing values and convert date columns.
Calculate daily and cumulative cases.
Find countries/states with the highest number of cases.
Calculate recovery and death rates using NumPy/Pandas.
Create:
Line graph of cases over time
Bar chart comparing countries/states
Histogram of daily cases
Seaborn heatmap showing correlations
Identify major trends in the data.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df2=pd.read_csv("covid19_big_dataset.csv")
df2
```

```python
df2.head()
```

```python
df2.shape
```

```python
df2.info()
```

```python
df2["Date"]=pd.to_datetime(df2["Date"])
df2.info()
```

```python
df2["Recovered"]=df2["Recovered"].astype("Int64")
```

```python
df2["Deaths"]=df2["Deaths"].astype("Int64")
```

```python
df2["Active"]=df2["Active"].astype("Int64")
```

```python
df2.info()
```

```python
df2.isnull().sum()
```

```python
df2.dropna(inplace=True)
df2
```

```python
df2.duplicated().sum()
```

```python
df2.drop_duplicates(inplace=True)
df2
```

```python
df2.sort_values("Date")
```

```python
country_daily = df2.groupby(["Date", "Country"])["Confirmed"].max().reset_index()
```

```python
country_daily
```

```python
country_daily["Daily_Cases"] = (
    country_daily.groupby("Country")["Confirmed"].diff().fillna(0)
)
```

```python
country_daily["Cumulative_Cases"] = (
    country_daily.groupby("Country")["Daily_Cases"].cumsum()
)
```

```python
country_daily.head(100)
```

```python
# Find Heighest Country with highest cases
country_cases = df2.groupby("Country")["Confirmed"].max().sort_values(ascending=False)
country_cases
```

```python
state_cases = df2.groupby(["Country", "State"])["Confirmed"].max().sort_values(ascending=False)
state_cases
```

```python
#Recovery Rate
total_recovered = df2["Recovered"].sum()
total_confirmed = df2["Confirmed"].sum()

recovery_rate = (total_recovered / total_confirmed) * 100

print("Recovery Rate:", recovery_rate)
```

```python
# Death rate
total_deaths = df2["Deaths"].sum()

death_rate = (total_deaths / total_confirmed) * 100

print("Death Rate:", death_rate)
```

```python
#line chart

daily_cases = df2.groupby("Date")["Confirmed"].sum()

plt.figure(figsize=(12,6))

plt.plot(daily_cases.index, daily_cases.values)

plt.title("COVID-19 Cases Over Time")
plt.xlabel("Date")
plt.ylabel("Confirmed Cases")

plt.show()
```

```python
# bar chart
top_countries = country_cases.head(10)

plt.figure(figsize=(12,6))

top_countries.plot(kind="bar")

plt.title("Top 10 Countries by COVID-19 Cases")
plt.xlabel("Country")
plt.ylabel("Confirmed Cases")

plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(12,4))

plt.hist(
    country_daily["Daily_Cases"],
    bins=100
)

plt.title("Distribution of Daily COVID-19 Cases")
plt.xlabel("Daily Cases")
plt.ylabel("Frequency")

plt.show()
```

```python
correlation = df2[["Confirmed", "Recovered", "Deaths", "Active"]].corr()
```

```python
#heatmap
plt.figure(figsize=(8,6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="Blues"
)

plt.title("COVID-19 Data Correlation Heatmap")

plt.show()
```
