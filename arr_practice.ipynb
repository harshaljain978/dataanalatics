```python
import numpy as np

# 1. Start with one array of names
names = np.array(["John Doe", "Jane Smith"])
print("Original:", names)

# 2. SPLIT: Break each name into [First, Last]
# (We use Python's built-in split here just to separate the strings)
first_names = np.array(["John", "Jane"])
last_names = np.array(["Doe", "Smith"])

# 3. MODIFY: Change just the first names
added_title = np.array(["Mr. John", "Ms. Jane"])

# 4. STACK: Glue the modified first names back to the last names
final_list = np.hstack((added_title, last_names))
print("\nStacked back together:\n", final_list)
```
