```python
# 1. What is Matplotlib?

# Matplotlib is a Python library used to create graphs and visualizations.

# You can use it to make:

#     Line charts
#     Bar charts
#     Scatter plots
#     Histograms
#     Pie charts
#     Subplots
#     Customized scientific/business visualizations

# The most important module is matplotlib.pyplot, which we usually import as plt:
```

```python
import matplotlib.pyplot as plt
```

```python
days = ["Monday","Tuesday","Wednessday","Thrusday","Friday"]
degree = [32,36,28,31,39]
```

```python
plt.plot(days,degree)
plt.title("Temprature Day-wise")
plt.xlabel("Days")
plt.ylabel("Degree")
plt.show()
```

```python
months=["Jan","Feb","Mar","Apr","May","Jun","jul","Aug","Sep","Oct","Nov","Dec"]
sales=[100,120,145,500,347,505,245,1000,645,1102,999,1300]
```

```python
plt.plot(months,sales)
plt.title("Sales Per-Month")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
plt.plot(months,sales,color="pink")
plt.title("Sales Per-Month")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
plt.plot(months,sales,color="pink",linewidth="10")
plt.title("Sales Per-Month")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
plt.plot(months,sales,color="pink",linewidth="4",linestyle="--")
plt.title("Sales Per-Month")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
plt.plot(months,sales,color="pink",linewidth="4",linestyle=":")
plt.title("Sales Per-Month")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
plt.plot(months,sales,color="pink",linewidth="3",linestyle="--",marker="o")
plt.title("Sales Per-Month")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
plt.plot(months,sales,color="pink",linewidth="3",linestyle="--",marker="s")
plt.title("Sales Per-Month")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
plt.plot(months,sales,color="pink",linewidth="1",linestyle="--",marker="^")
plt.title("Sales Per-Month")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
x=[1,2,3,4,5]
x1=[2,4,5,8,9]
x2=[5,3,7,10,11]

plt.plot(x,x1,color="orange",label="2x",linestyle=":")
plt.plot(x1,x2,color="#823420",label="3x",linestyle="--",marker="o")
plt.plot(x2,x,color="blue")

plt.legend()
```

```python
x=[10,20,40,60,70,80,100]
y=[1000,2000,4000,7000,8000,9000,100000]
```

```python
student=["Salim","Suresh","Harshal","Priyal","Aditya"]
marks=[72,82,44,77,92]
```

```python
plt.bar(student, marks, color='brown', edgecolor='black', linewidth=2,linestyle=":")
plt.scatter(student, marks,color='Pink',linewidth=10)
plt.title('Students Marks out of 100')
plt.xlabel('Student Name')
plt.ylabel('Marks')
plt.grid(True)
plt.show()
```

```python
TouristCitiesofIndia=["Jodhpur","Udaipur","Jaipur","Jaisalmer","Pushkar"]
NumberofTouristperYear = [120000,342000,201000,256790,118980]
plt.plot(TouristCitiesofIndia,NumberofTouristperYear,color="brown",linewidth=1,marker="o",label="Number of Touristper Year")
plt.bar(TouristCitiesofIndia,NumberofTouristperYear,color="pink",linewidth=1,linestyle="--",edgecolor="Brown",width=0.3,label="Number of Touristper Year")
plt.xlabel("Tourist Cities of India")
plt.ylabel("Number of Tourist per Year")
plt.title("Footfall in Top Rajesthan Cities")
plt.grid()
plt.legend()
plt.show()
```
