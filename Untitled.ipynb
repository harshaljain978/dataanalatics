```python
import numpy as np
```

```python
names = np.array(["Aarav", "Priya", "Rohan", "Neha"])
ages = np.array([14, 16, 15, 17])
math_marks = np.array([85, 90, 78, 92])
science_marks = np.array([88, 76, 95, 89])
```

```python
print("Arrays initialized successfully!")
```

```python
#Calculate Average Math Marks
avg_math = math_marks.mean()
print(f"Average Math Mark: {avg_math}")
```

```python
#Find the Highest Science Mark and Who Scored It
highest_science = science_marks.max()
top_student_index = science_marks.argmax()
top_student = names[top_student_index]

print(f"Highest Science Mark: {highest_science} (Scored by {top_student})")
```

```python
#Filter Students Older Than 15
age_mask = ages > 15
older_students = names[age_mask]
there_ages = ages[age_mask]
```

```python
print("Students older than 15:")
for name, age in zip(older_students, there_ages):
    print(f"{name}: {age} Years Old")
```
