```python
import  numpy as pd
import pandas as pd
```

```python
employee = pd.DataFrame({
    "employee_ID":[1,2,3,6,7],
    "name":["John","Anna","Peter","Linda","Bob"],
    "Department":["HR","IT","Finance","IT","HR"]
})
```

```python
salary=pd.DataFrame({
    "employee_ID":[1,2,3,4,5],
    "salary":[100,200,300,400,500],
    "bonus":[50,60,70,80,90]
})
```

```python
employee
```

```python
salary
```

```python
pd.merge(employee,salary,on="employee_ID",how="inner")
#inenr is used to get the common or the instersection of the tables
```

```python
pd.merge(employee,salary,on="employee_ID",how="outer")
#outer is used to take all the vales in both the tables and put the rest values a NaN
```

```python
pd.merge(employee,salary,on="employee_ID",how="right")
```

```python
pd.merge(employee,salary,on="employee_ID",how="left")
```

```python
import numpy as np
data = {
    'Date': pd.date_range("2024-01-01",periods=20),
    'Product':['A','B','C','D'] * 5,
    'Region': ['East','West','North','South','East','West','North','South','East','West',
               'North','South','East','West','North','South','East','West','North','South'
              ],
    'sales':np.random.randint(100,1000,20),
    'Units': np.random.randint(10,100,20),
    'Rep': ['Priyal','Harshal','Aditya','Akshay','Kashish','Arya','Aishwarya','Urvish','Shivani','Sakshi','Shraddha','samiya','shravani',"saurabh",'bob',
           'linda','john','motu','patlu','doremon'
           ]
}
```

```python
df1 = pd.DataFrame(data)
```

```python
df1
```

```python
#Pivot Table
```

```python
pd.pivot_table(df1,values="sales",index="Region",columns="Product",aggfunc="median")
```

```python
pd.pivot_table(df1,values=["sales","Units"],index="Region",columns="Product")
```
